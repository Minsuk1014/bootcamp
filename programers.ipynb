```python
import datetime # 시간 정보 활용

print(f"--- Iterable 예제 (현재 시각: {datetime.datetime.now().strftime('%H:%M:%S KST')}) ---") # 오전 11:52

# 다양한 Iterable 객체들
my_list = [10, 20, 30]
my_tuple = ('월', '화', '수')
my_string = "Seoul" # 위치 정보 활용
my_dict = {'a': 1, 'b': 2} # 기본적으로 key에 대해 iterable
my_set = {1.1, 2.2, 3.3}
my_range = range(3)

# iter() 함수로 이터레이터 얻기 확인
print(f"리스트로부터 이터레이터 얻기: {iter(my_list)}")
print(f"문자열로부터 이터레이터 얻기: {iter(my_string)}")
print(f"딕셔너리로부터 이터레이터 얻기: {iter(my_dict)}") # key iterator 반환
print(f"셋으로부터 이터레이터 얻기: {iter(my_set)}")
print(f"range로부터 이터레이터 얻기: {iter(my_range)}")
```

```python
print("\n--- 제너레이터 표현식 예제 ---")

# 0~9 제곱수 제너레이터 표현식
squares_genex = (x * x for x in range(10))
print(f"제너레이터 표현식 객체: {squares_genex}") # <generator object <genexpr> at ...>

# 제너레이터 표현식 순회 (필요할 때 값이 계산됨)
print("제곱수 순회:")
for sq in squares_genex:
    print(f"  제곱: {sq}")

# 제너레이터 표현식은 한 번만 순회 가능 (소진됨)
print(f"두 번째 순회 시도: {list(squares_genex)}") # 출력: [] (빈 리스트)

# 함수 인자로 바로 사용 (중간 리스트 생성 안 함)
total_sum_of_evens = sum(n for n in range(1, 11) if n % 2 == 0) # 1~10 중 짝수의 합
print(f"짝수 합 (제너레이터 표현식): {total_sum_of_evens}") # 출력: 30 (2+4+6+8+10)
```
