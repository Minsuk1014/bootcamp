```python
import requests
import pymysql
#문제를 똑바로 보자
"""

"""

def stock_to_sql(symbol,start,end):
    data = eval(requests.get(f"https://m.stock.naver.com/front-api/external/chart/domestic/info?symbol={symbol}&requestType=1&startTime={start}&endTime={end}&timeframe=day").text.strip())

    conn = pymysql.connect(host='127.0.0.1', user = 'play', passwd = '123', database= 'sk15', port=3306) # 기본포트 3306로 연결
    cur = conn.cursor() # cursor로 가리키기기
    add_table = """CREATE TABLE naver_day_stock (
	st_code INT not NULL,
    st_date DATE NOT NULL,
	st_high INT NULL DEFAULT NULL,
	st_low INT NULL DEFAULT NULL,
	st_close INT NULL DEFAULT NULL,
	st_traed INT NULL DEFAULT NULL,
	st_expend FLOAT NULL DEFAULT NULL,
	PRIMARY KEY (st_code, st_date) )"""

    cur.execute(add_table)
    sql = "INSERT INTO naver_day_stock VALUES (%s, %s, %s, %s, %s, %s, %s)" # 
    for x in data[1:] :
        cur.execute(sql, [symbol,x[:-1]]) # 넣기
    cur.execute("TRUNCATE TABLE naver_day_stock;")
    conn.commit()

stock_to_sql(int("000885"),20250301,20250305)
```
