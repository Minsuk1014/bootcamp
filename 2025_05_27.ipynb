```python
def fibonacci_recursive(n: int) -> int:
    """
    n번째 피보나치 수를 재귀적으로 계산하여 반환합니다.
    (F(0)=0, F(1)=1 기준)

    주의: 이 순수 재귀 방식은 n이 커질수록 동일한 계산을 반복하여
          매우 비효율적입니다 (시간 복잡도 O(2^n)).
          실제 사용 시에는 메모이제이션(동적 프로그래밍)을 사용한 최적화가 필요합니다.

    재귀함수는 기본적으로 스택 구조처럼 이동.
    """
    if not isinstance(n, int) or n < 0: # n이 인트인가? isinstance(n,int)
        raise ValueError("입력값은 0 이상의 정수여야 합니다.") # 에러가 보이도록 만들기
    if n <= 1:  # 기본 조건: F(0) = 0, F(1) = 1
        return n
    else:
        # F(n) = F(n-1) + F(n-2)
        return fibonacci_recursive(n - 1) + fibonacci_recursive(n - 2)
    
fibonacci_recursive(3)

help(fibonacci_recursive)
```

```python
# 재귀 버전 테스트
if __name__ == "__main__":
    print("재귀 버전 피보나치 수열 (n번째 수):")
    for i in range(10): # 0부터 9번째 피보나치 수까지 출력
        # n이 커지면 매우 느려지므로 작은 범위에서 테스트
        if i < 20: # 실제로는 15-20만 넘어가도 눈에 띄게 느려짐
             print(f"F({i}) = {fibonacci_recursive(i)}")
        else:
            print(f"F({i}) 계산은 재귀 방식으로는 매우 오래 걸릴 수 있습니다.")
            break
    
    # 예: F(10) 계산
    # print(f"\nF(10) = {fibonacci_recursive(10)}") # 55
    # print(f"F(30) = {fibonacci_recursive(30)}") # 이 계산은 상당히 오래 걸립니다!
```
